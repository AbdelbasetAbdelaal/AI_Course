```python
import mysql.connector

# Database connection configuration for XAMPP
# Default XAMPP settings: user='root', password='' (empty)
db_config = {
    'host': '127.0.0.1',
    'user': 'root',
    'password': '',
    'database': 'school_db'
}

def fetch_student_data():
    try:
        # Establish the connection
        connection = mysql.connector.connect(**db_config)
        
        if connection.is_connected():
            cursor = connection.cursor()
            
            # Execute the SELECT query
            query = "SELECT student_id, first_name, last_name, grade_level, email FROM students"
            cursor.execute(query)
            
            # Fetch all rows from the result set
            records = cursor.fetchall()
            
            print(f"Total students found: {cursor.rowcount}\n")
            
            # Print table headers
            print(f"{'ID':<5} | {'First Name':<12} | {'Last Name':<12} | {'Grade':<8} | {'Email'}")
            print("-" * 75)
            
            # Loop through the records and print them
            for row in records:
                print(f"{row[0]:<5} | {row[1]:<12} | {row[2]:<12} | {row[3]:<8} | {row[4]}")

    except mysql.connector.Error as e:
        print(f"Error while connecting to MySQL: {e}")
        
    finally:
        # Close the connection and cursor
        if 'connection' in locals() and connection.is_connected():
            cursor.close()
            connection.close()
            print("\nMySQL connection is closed.")

if __name__ == "__main__":
    fetch_student_data()
```
