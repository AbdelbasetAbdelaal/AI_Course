```python
from abc import ABC, abstractmethod 
class LibraryItem(ABC):
    def __init__(self,title,ID):
        self.title = title
        self.ID = ID
        self.is_borrowed = False
    @abstractmethod
    def get_loan_period(self):

        pass
    def __str__(self):
        status = "Borrowed" 
        if self.is_borrowed:
            return status
        else:
            status = "available" 
            return status       
```
