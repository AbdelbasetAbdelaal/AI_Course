```python
class BankAccount:
    """
    A class representing a bank account with deposit and withdrawal capabilities.
    """

    def __init__(self, owner, balance=0.0):
        self.owner = owner
        self.__balance = balance  # Private attribute for encapsulation

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"\n[Deposit Success]: ${amount:,.2f} added.")
            self.display_balance()
        else:
            print("\n[Error]: Deposit amount must be a positive number.")

    def withdraw(self, amount):
        """Subtracts money if the amount is valid and funds are available."""
        if amount <= 0:
            print("\n[Error]: Withdrawal amount must be positive.")
        elif amount > self.__balance:
            print(f"\n[Error]: Insufficient funds! You only have ${self.__balance:,.2f}")
        else:
            self.__balance -= amount
            print(f"\n[Withdrawal Success]: ${amount:,.2f} removed.")
            self.display_balance()

    def display_balance(self):
        print(f"Current balance for {self.owner}: ${self.__balance:,.2f}")

    def __str__(self):
        return f"\n--- Account Details ---\nOwner: {self.owner}\nBalance: ${self.__balance:,.2f}\n-----------------------"


def run_bank_system():
    print("Welcome to the Python Bank System")
    
    # Create the account
    name = input("Enter the account holder's name: ")
    try:
        initial = float(input("Enter initial deposit: "))
    except ValueError:
        initial = 0.0
        
    account = BankAccount(name, initial)
    print(account)

    # Interactive Menu
    while True:
        print("\nWhat would you like to do?")
        print("1. Deposit Money")
        print("2. Withdraw Money")
        print("3. Check Balance")
        print("4. Exit")
        
        choice = input("Select an option (1-4): ")

        if choice == '1':
            try:
                amount = float(input("Enter deposit amount: "))
                account.deposit(amount)
            except ValueError:
                print("Invalid input. Please enter a number.")

        elif choice == '2':
            try:
                amount = float(input("Enter withdrawal amount: "))
                account.withdraw(amount)
            except ValueError:
                print("Invalid input. Please enter a number.")

        elif choice == '3':
            account.display_balance()

        elif choice == '4':
            print(f"Goodbye, {account.owner}!")
            break
        else:
            print("Invalid choice, please try again.")

if __name__ == "__main__":
    run_bank_system()
```
